```python
#Identifying the missing value
df.isnull().sum()
```

```python
#Drop the cabin col
df = df.drop(["Cabin"],axis=1)
```

```python
df.isnull().sum()
```

```python
#Handling the missing value for Age
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
df.isnull().sum()
```

```python
#Missing value handling for embarked column
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```
