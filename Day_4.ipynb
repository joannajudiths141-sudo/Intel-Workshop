```python
iris_features = iris[
    [
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ]
].copy()

iris_features.head()
```

```python
import matplotlib.pyplot as plt
iris_features.hist(
    bins=15,
    figsize=(10, 5)
)

plt.suptitle("Iris Feature Distribution")
plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
iris_scaler = StandardScaler()

iris_scaled = iris_scaler.fit_transform(
    iris_features
)
```

```python
from sklearn.cluster import KMeans

inertia_values = []

for k in range(1, 11):
  model = KMeans(
      n_clusters=k,
      random_state=42,
      n_init=10
  )

  model.fit(iris_scaled)
  inertia_values.append(model.inertia_)
```

```python
plt.figure(figsize=(7, 4))

plt.plot(
    range(1, 11),
    inertia_values,
    marker='o',
)

plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.xticks(range(1, 11)),
plt.tight_layout()
plt.show()
```

```python
Kmeans_model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

iris_clusters = Kmeans_model.fit_predict(
    iris_scaled
)
```

```python
iris_clustered = iris.copy()

iris_clustered["Cluster"] = iris_clusters

iris_clustered.head()
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(
    iris_clustered["petal_length"],
    iris_clustered['petal_width'],
    c=iris_clustered["Cluster"],
)

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("K-Means Clustering on Iris Data")
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(
    iris_scaled,
    iris_clusters
)

print("Silhouette Score:", score)
```
